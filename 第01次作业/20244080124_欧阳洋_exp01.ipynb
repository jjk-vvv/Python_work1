```python
operations = [("8 / 3", 8 / 3), ("8 // 3", 8 // 3),
              ("8 % 3", 8 % 3), ("2 ** 7", 2 ** 7)]
for expression, value in operations:
    print(f"{expression} = {value}, type = {type(value).__name__}")

basic_values = [42, 3.14, "Python", True]
for value in basic_values:
    print(f"value = {value!r}, type = {type(value).__name__}")
```

### 1.2 类型转换与字符串操作

检查三种合法转换并捕获非法整数转换；对同一字符串分别执行去空白、大小写转换、切片、拆分和连接。

## 三、实验 2：函数、条件、循环与递归

### 2.1 成绩分类与边界验证

先检查成绩是否在 `[0, 100]`，再从高分到低分判断等级。
用 `0、59、60、79、80、89、90、100` 验证端点和分段边界，
用 `59.5、79.5、89.5` 验证“不足”对应的区间，并测试范围外的 `-1、101`。

```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("分数必须在 0 到 100 之间")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"

for score in [0, 59, 59.5, 60, 79, 79.5, 80, 89, 89.5, 90, 100]:
    print(f"{score:5g} 分：{score_level(score)}")
for score in [-1, 101]:
    try:
        score_level(score)
    except ValueError as error:
        print(f"{score:3d} 分：已捕获 {type(error).__name__}，{error}")
```

**结果分析：** `0 ≤ score < 60` 为不及格，`60 ≤ score < 80` 为及格，
`80 ≤ score < 90` 为良好，`90 ≤ score ≤ 100` 为优秀。
分段边界 `60、80、90` 都归入较高等级，端点 `0、100` 属于有效范围。
范围检查位于等级判断之前，避免将 `101` 误判为优秀或将负数误判为不及格。

### 2.2 两种循环与两种阶乘实现

`for` 和 `while` 都包含 `1～100` 的所有整数。阶乘输入限定为非负整数，
递归版以 `0! = 1` 为终止条件，迭代版从 `1` 开始逐项相乘。

```python
sum_for = 0
for number in range(1, 101):
    sum_for += number

sum_while = 0
number = 1
while number <= 100:
    sum_while += number
    number += 1
print(f"for 求和 = {sum_for}，while 求和 = {sum_while}，结果相等：{sum_for == sum_while}")

def factorial_recursive(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n 必须是非负整数")
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n 必须是非负整数")
    result = 1
    for value in range(2, n + 1):
        result *= value
    return result

for n in [0, 1, 5, 10]:
    recursive = factorial_recursive(n)
    iterative = factorial_iterative(n)
    print(f"{n}!：递归 = {recursive}，迭代 = {iterative}，相等：{recursive == iterative}")
```

**结果分析：** 两种循环均得到 `5050`，与 `100 × 101 / 2` 一致。
`range(1, 101)` 的终点 `101` 不包含在序列中；`while` 通过 `<= 100` 包含最后一项。
`0!、1!、5!、10!` 分别为 `1、1、120、3628800`，两种阶乘实现结果一致。
递归会为每次调用保留调用栈，迭代用循环累计结果；这里比较的是实现和结果，没有进行运行时间测量。

### 2.3 统计函数、元组解包与可变参数

`summary()` 接收非空数值列表，一次返回最小值、最大值和平均值；调用处用三个变量解包。
`sum_all(*args)` 将任意数量的参数收集为元组，再求和。

```python
def summary(values):
    if not values:
        raise ValueError("values 必须是非空数值列表")
    return min(values), max(values), sum(values) / len(values)

minimum, maximum, average = summary([85, 58, 92, 76, 60, 89])
print(f"最小值 = {minimum}，最大值 = {maximum}，平均值 = {average:.2f}")

def sum_all(*args):
    return sum(args)

print("无参数：", sum_all())
print("单个参数：", sum_all(5))
print("多个参数：", sum_all(1, 2, 3, 4))
```

**结果分析：** 三项统计结果为 `58、92、76.67`；函数返回的是包含三个元素的元组。
平均值计算使用全部六个分数，显示时用 `.2f` 保留两位小数。
无参数时 `args` 是空元组，`sum()` 返回 `0`；单参数和多参数示例分别得到 `5、10`。
统计函数要求非空输入，显式检查可以给出清楚的错误原因。

## 四、实验 3：列表、字典、元组与基础统计

### 3.1 组织成绩数据与计算统计指标

使用题目规定的六个姓名和分数，通过 `zip()` 建立对应关系；
`items()` 遍历字典，`enumerate(..., start=1)` 生成从 1 开始的序号。
用列表推导式筛选及格者并计算成绩平方，及格率以 `score >= 60` 为标准。

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]
score_by_name = dict(zip(names, scores))

print("姓名到成绩的字典：", score_by_name)
print("items() 遍历：")
for name, score in score_by_name.items():
    print(f"{name}: {score}")
print("enumerate() 编号：")
for index, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"{index}. {name}: {score}")

passed_names = [name for name, score in zip(names, scores) if score >= 60]
squared_scores = [score ** 2 for score in scores]
count = len(scores)
average_score = sum(scores) / count
highest_score = max(scores)
lowest_score = min(scores)
pass_rate = len(passed_names) / count
print("及格学生：", passed_names)
print("成绩平方：", squared_scores)
print(f"人数 = {count}，平均分 = {average_score:.2f}，最高分 = {highest_score}，"
      f"最低分 = {lowest_score}，及格率 = {pass_rate:.2%}")
```

**结果分析：** 六人成绩总和为 `460`，均分为 `460 / 6 ≈ 76.67`；最高分为 `92`，最低分为 `58`。
除 Bob 外其余五人及格，Eva 的 `60` 分计入及格，及格率为 `5 / 6 ≈ 83.33%`。
字典适合按姓名查分，列表适合按顺序处理数据；列表推导式保留了原数据的遍历顺序。

### 4.2 逐行解析与异常记录

解析函数先检查逗号分隔后的字段数和姓名，再把成绩转换为整数并检查 `[0, 100]` 范围。
读取时跳过第 1 行表头，`enumerate(..., start=2)` 使错误行号与文本文件一致。
每行单独捕获 `ValueError`，一行失败不会中断后续记录处理。

```python
def parse_student(line):
    """将 ``name,score`` 文本解析为包含 name 与 score 的字典。

    字段数量、姓名、分数格式和 0--100 的分数范围不符合要求时抛出 ValueError。
    """
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError("记录必须包含姓名和成绩两个字段")

    name = parts[0].strip()
    score_text = parts[1].strip()
    if not name:
        raise ValueError("姓名不能为空")

    try:
        score = int(score_text)
    except ValueError as error:
        raise ValueError("成绩必须是整数") from error

    if not 0 <= score <= 100:
        raise ValueError("成绩必须在 0 到 100 之间")

    return {"name": name, "score": score}

valid_records = []
invalid_records = []
with open(students_path, encoding="utf-8") as file:
    header = next(file).strip()
    assert header == "name,score", "输入文件表头与题目不一致"
    for line_number, line in enumerate(file, start=2):
        try:
            valid_records.append(parse_student(line))
        except ValueError as error:
            invalid_records.append((line_number, line.strip(), str(error)))

print("有效记录：")
for record in valid_records:
    print(record)
print("无效记录：")
for line_number, raw_line, reason in invalid_records:
    print(f"第 {line_number} 行 {raw_line!r}：{reason}")
```

**结果分析：** 文件包含六条有效记录和三条无效记录。

| 文件行号 | 原始内容 | 无效原因 |
|---:|---|---|
| 8 | `Grace,abc` | 成绩不能转换为整数 |
| 9 | `Henry,105` | 成绩超出 0～100 |
| 10 | `Incomplete` | 缺少成绩字段 |

这些行号包含表头行，而不是从第一条学生记录开始计数。无效记录没有参与后面的统计。

```python
average_valid = sum(item["score"] for item in valid_records) / len(valid_records)
pass_rate_valid = sum(item["score"] >= 60 for item in valid_records) / len(valid_records)
report_body = (
    f"有效记录数: {len(valid_records)}\n"
    f"无效记录数: {len(invalid_records)}\n"
    f"平均分: {average_valid:.2f}\n"
    f"及格率: {pass_rate_valid:.2%}\n"
)
with open(report_path, "w", encoding="utf-8") as file:
    file.write(report_body)
print("w 写入后的内容：")
with open(report_path, encoding="utf-8") as file:
    print(file.read(), end="")

with open(report_path, "a", encoding="utf-8") as file:
    file.write("无效记录详情:\n")
    for line_number, raw_line, reason in invalid_records:
        file.write(f"第 {line_number} 行 {raw_line!r}: {reason}\n")
    file.write("说明：w 模式覆盖已有内容；a 模式在文件末尾追加内容。\n")
print("a 追加后的完整内容：")
with open(report_path, encoding="utf-8") as file:
    print(file.read(), end="")
```

**结果分析：** `w` 打开已有文件时会截断旧内容，`a` 会保留已有内容并在末尾写入。

### 5.1 学生类与研究生子类

`Student` 用属性保存姓名和分数，`is_passed()` 判断是否达到 60 分，`__str__()` 定义显示形式。
按实验 4 的有效记录创建对象。`GraduateStudent` 通过 `super()` 初始化公共属性，
再增加研究主题并重写显示方法。

```python
import copy

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"

class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"research_topic={self.research_topic})")

students = [Student(item["name"], item["score"]) for item in valid_records]
for student in students:
    print(student, "及格：", student.is_passed())
graduate = GraduateStudent("Carol", 92, "Data Analysis")
print(graduate)
print("研究生对象属于 Student：", isinstance(graduate, Student))
print("继承的及格判断：", graduate.is_passed())
```

**结果分析：** 六条有效记录对应六个学生对象，及格判断与实验 3 的五人及格结果一致。
子类实例继承 `is_passed()`，重写的 `__str__()` 额外显示 `research_topic`。
`super().__init__()` 复用父类对姓名和分数的初始化，避免重复定义公共属性。

### 5.2 赋值、浅拷贝与深拷贝

三次比较都从新的 `{"name": "Alice", "scores": [85, 90]}` 开始。
修改复制对象中的嵌套列表，分别检查外层字典和内层列表是否是同一对象。

```python
original_assigned = {"name": "Alice", "scores": [85, 90]}
assigned = original_assigned
print("直接赋值，修改前：", original_assigned, assigned)
assigned["scores"][0] = 100
print("直接赋值，修改后：", original_assigned, assigned)
print("外层相同：", original_assigned is assigned,
      "内层相同：", original_assigned["scores"] is assigned["scores"])

original_shallow = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(original_shallow)
print("浅拷贝，修改前：", original_shallow, shallow)
shallow["scores"][0] = 100
print("浅拷贝，修改后：", original_shallow, shallow)
print("外层相同：", original_shallow is shallow,
      "内层相同：", original_shallow["scores"] is shallow["scores"])

original_deep = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(original_deep)
print("深拷贝，修改前：", original_deep, deep)
deep["scores"][0] = 100
print("深拷贝，修改后：", original_deep, deep)
print("外层相同：", original_deep is deep,
      "内层相同：", original_deep["scores"] is deep["scores"])
```

**结果分析：**

| 方式 | 外层字典为同一对象 | 内层成绩列表为同一对象 | 修改副本的内层列表后，原对象首项 |
|---|---|---|---:|
| 直接赋值 | 是 | 是 | 100 |
| `copy.copy()` | 否 | 是 | 100 |
| `copy.deepcopy()` | 否 | 否 | 85 |

直接赋值增加了同一对象的引用。浅拷贝创建新的外层字典，但仍共享嵌套列表；
深拷贝同时复制本例中的嵌套列表，因此修改副本不会改变原对象。
是否相等反映内容，`is` 判断是否为同一对象，两种判断的含义不同。
